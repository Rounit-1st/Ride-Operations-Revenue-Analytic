#Feature Engineering

```python
# Create Ride Date
rides['ride_date'] = rides['ride_start_time'].dt.date
```

```python
rides[['ride_start_time', 'ride_date']].head()
```

```python
# Extract Ride Hour
rides['ride_hour'] = rides['ride_start_time'].dt.hour
```

```python
rides[['ride_start_time', 'ride_hour']].head()
```

```python
# Day of Week
rides['day_name'] = rides['ride_start_time'].dt.day_name()
```

```python
rides[['ride_start_time', 'day_name']].head()
```

```python
# Month Name
rides['month'] = rides['ride_start_time'].dt.month_name()
```

```python
rides[['ride_start_time', 'month']].head()
```

```python
# Quarter
rides['quarter'] = rides['ride_start_time'].dt.quarter
```

```python
# Year
rides['year'] = rides['ride_start_time'].dt.year
```

```python
# Weekend Flag
rides['weekend_flag'] = rides['day_name'].isin(['Saturday', 'Sunday'])
```

```python
rides['weekend_flag'].value_counts()
```

```python
# Time sot
def time_slot(hour):
    if 0 <= hour < 6:
        return 'Late Night'
    elif 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 22:
        return 'Evening'
    else:
        return 'Night'
```

```python
rides['time_slot'] = rides['ride_hour'].apply(time_slot)
```

```python
rides['time_slot'].value_counts()
```

```python
# Fare per Km
rides['fare_per_km'] = (
    rides['fare_amount'] / rides['distance_km']
).round(2)
```

```python
rides[['fare_amount', 'distance_km', 'fare_per_km']].head()
```

```python
#Trip category
def trip_category(distance):
    if distance < 5:
        return 'Short'
    elif distance <= 15:
        return 'Medium'
    else:
        return 'Long'
```

```python
rides['trip_category'] = rides['distance_km'].apply(trip_category)
```

```python
rides['trip_category'].value_counts()
```

```python
rides.head()
```

```python
rides['month_number'] = rides['ride_start_time'].dt.month
rides['month_name'] = rides['ride_start_time'].dt.month_name()
```
